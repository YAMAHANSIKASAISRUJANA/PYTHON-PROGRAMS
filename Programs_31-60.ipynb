```python
#41. Sum of Prime Numbers


n = 20
total = 0

for num in range(2, n + 1):
    count = 0

    for i in range(1, num + 1):
        if num % i == 0:
            count = count + 1

    if count == 2:
        total = total + num

print("Sum:", total)
```

```python
#43. Armstrong Number


n = 153
temp = n
total = 0
digits = len(str(n))

while temp > 0:
    digit = temp % 10
    total = total + digit ** digits
    temp = temp // 10

if total == n:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

```python
#44. Armstrong Numbers from 1 to N


n = 500

for num in range(1, n + 1):
    temp = num
    total = 0
    digits = len(str(num))

    while temp > 0:
        digit = temp % 10
        total = total + digit ** digits
        temp = temp // 10

    if total == num:
        print(num)
```

```python
#47. Strong Number


n = 145
temp = n
total = 0

while temp > 0:
    digit = temp % 10
    factorial = 1

    for i in range(1, digit + 1):
        factorial = factorial * i

    total = total + factorial
    temp = temp // 10

if total == n:
    print("Strong number")
else:
    print("Not a Strong number")
```

```python
#48. Strong Numbers from 1 to N


n = 500

for num in range(1, n + 1):
    temp = num
    total = 0

    while temp > 0:
        digit = temp % 10
        factorial = 1

        for i in range(1, digit + 1):
            factorial = factorial * i

        total = total + factorial
        temp = temp // 10

    if total == num:
        print(num)
```
